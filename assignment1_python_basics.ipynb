```python
# 1. Simple ATM Calculator
# Write a program that takes the user's account balance and withdrawal amount. If
# sufficient balance is available, deduct the amount and display the updated balance.
# Otherwise, print an error message.
# Sample Input:
# Balance
# =
# Withdrawal  =
# Sample Output:
# 5000
# 1500
# Transaction successful. Remaining balance: ₹3500

balance=eval(input("enter the balance amount"))
withdrwal=eval(input("enter the withdrwal amount"))
if balance>0 and balance>withdrwal:
  print("Transaction Successful.Remaining Balance:",balance-withdrwal)
else:
  print("Transaction Not Successful")
```

```python
# 2. Electricity Bill Generator
# Create a bill calculator where the user enters the number of units consumed. The rate is
# ₹5 per unit for the first 100 units, ₹7 for the next 100, and ₹10 for anything above 200
# units. Calculate and display the total bill.
# Sample Input:
# Units consumed
# Sample Output:
# =
# 275
# Bill amount = ₹1000 (100×5 + 100×7 + 75×10)

units=eval(input("Enter the units"))
bill=0

if units<=100:
  bill=units*5
elif units<=200:
  bill=(5*100)+((units-100)*7)
else:
  bill=(5*100)+(7*100)+((units-200)*10)
print("Total Bill = ₹", bill)
```

```python
# 3. Temperature Converter
# Write a Python program to convert a given temperature value between Celsius and
# Fahrenheit. Accept input as a number and a unit indicator ('C' or 'F').
# Sample Input:
# Temp
# Unit
# =
# =
# Sample Output:
# "37"
# "C"
# Temperature in Fahrenheit: 98.6°F

temperature=eval(input("Enter the temperature"))
unit=input("Enter the unit (C/F):")

if unit=="C":
  f=(temperature*9/5)+32
  print("Temperature in Fahrenheit",f,"°F")
elif unit=="F":
  c=(temperature-32)*5/9
  print("Temperature in Celsius",c,"°C")
else:
  print("Invalid")



```

```python
# 4. Mobile Recharge Validity
# Write a program that takes the recharge date and today’s date as input and checks
# whether the 28-day recharge plan is still valid or expired.
# Sample Input:
# Recharge
# Today :
# date: 2025-07-01
# 2025-07-30
# Sample Output:
# Plan expired. Recharge again.

# rechargedate=input("Enter the recharge date(yyyy-mm-dd)")
# todaydate=input("Enter the today date(yyyy-mm-dd)")
# r=int(rechargedate[8:10])
# t=int(todaydate[8:10])
# date=t-r
# if date<=28:
#   print("Plan is valid")
# else:
#   print("Plan expired.Recharge again")

from datetime import datetime

recharge_date = input("Enter recharge date")
today_date = input("Enter today's date")

recharge = datetime.strptime(recharge_date, "%Y-%m-%d")
today = datetime.strptime(today_date, "%Y-%m-%d")

days = (today - recharge).days

if days <= 28:
    print("Plan is still valid.")
else:
    print("Plan expired. Recharge again.")


```

```python
# 6. Sum of even numbers
# Write a program that takes a number as input and prints the sum of all even numbers
# from 1 to that number.
# Sample Input:
# Number: 10
# Sample Output:
# Sum of even numbers: 30
# (2 + 4 + 6 + 8 + 10)

n=eval(input("Enter the Number"))
sum=0
for i in range(1,n+1):
  if i%2==0:
    sum=sum+i
print(sum)
```

```python
# 8. Dictionary – Word Frequency
# Write a program that takes a sentence and displays the frequency of each word using a
# dictionary.
# Sample Input:
# Sentence: hello world hello python
# Sample Output:
# hello: 2
# world: 1
# python: 1

sentence = input("Enter the sentence: ")
words = sentence.split()
frequency = {}
for word in words:
    if word in frequency:
        frequency[word] = frequency[word] + 1
    else:
        frequency[word] = 1

for word in frequency:
    print(word, ":", frequency[word])
```

```python
# 8. Functions – Check Prime
# Sample Input:
# Enter a number
# Sample output:
# 17 is a prime number.

def check(n):

  if n<=1:
    print("Not Prime")
  for i in range(2,n):
    if(n%i==0):
      print("Not Prime")
  else:
    print("Prime")
n = int(input("Enter the number: "))
check(n)
```

```python
# 10. Count Divisible Numbers
# Write a program that takes two numbers as input: a range limit N and a divisor D. Count
# how many numbers between 1 and N are divisible by D.
# Sample Input:
# Enter the range limit (N): 50
# Enter the divisor (D): 5
# Sample output:
# Count of numbers divisible by 5: 10
# (i.e., 5, 10, 15, ..., 50)
def check(a,b):
  return len([i for i in range(1,a+1) if i%b==0])
n=eval(input("enter the range limit"))
d=eval(input("enter the diviisor"))
check(n,d)
```
